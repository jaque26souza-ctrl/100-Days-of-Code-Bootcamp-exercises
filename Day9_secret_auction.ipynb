#Welcome to Blind Auction! | Bem-vindo ao Leilão às Cegas!


*   Aula 9 do bootcamp
*   Criar um programa de leilão às cegas
*   Dicionários e Nesting






```python
logo = r'''
                         ___________
                         \         /
                          )_______(
                          |"""""""|_.-._,.---------.,_.-._
                          |       | | |               | | ''-.
                          |       |_| |_             _| |_..-'
                          |_______| '-' `'---------'` '-'
                          )"""""""(
                         /_________\\
                       .-------------.
                      /_______________\\
'''
print(logo)
print("# Welcome to Blind Auction! | Bem vindo ao Leilão às Cegas!")
print("* Here you can make a blind auction with your colleagues or friends. | Aqui você pode criar um leilão às cegas com seus colegas ou amigos.")
print("* No one knows the bids of each other. | Ninguém sabe quais são os lances uns dos outros.")
print("* In the end, the program announces the winner!\n")

bids = {}
continue_bidding = True

def find_highest_bidder(bidding_dictionary):
    highest_bid = 0
    winner = ""
    for bidder in bidding_dictionary:
        bid_amount = bidding_dictionary[bidder]
        if bid_amount > highest_bid:
            highest_bid = bid_amount
            winner = bidder
    print(f"\nThe winner is {winner} with a bid of ${highest_bid}.")

while continue_bidding:
    name = input("What is your name?: ")
    price = int(input("What is your bid?: $"))

    bids[name] = price

    should_continue = input("Are there any other bidders? Type 'yes' or 'no'.\n").lower()
    if should_continue == "no":
        continue_bidding = False
        find_highest_bidder(bids)
    elif should_continue == "yes":
        print("\n" * 150)  # Limpa a tela simulando a rolagem para o próximo colega não ver o lance
```
