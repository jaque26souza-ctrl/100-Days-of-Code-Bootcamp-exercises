#Jogo da Forca - Hangman

```python
# @title 🎮 CLIQUE AQUI PARA JOGAR

import random

BANNER = """
====================================================================
                        J O G O   D A   F O R C A
====================================================================
"""

DEMO_TAG = """
--------------------------------------------------------------------
                       [ HANGMAN DEMO VERSION ]
--------------------------------------------------------------------
"""

# Dicionário com categorias e sublistas de palavras
categorias = {
    "Tecnologia": ["DESENVOLVIMENTO", "PYTHON", "ALGORITMO", "COMPUTADOR", "CELULAR"],
    "Animais": ["GATO", "CACHORRO", "LAGARTO", "BORBOLETA", "ELEFANTE", "RAPOSA", "GOLFINHO", "BALEIA", "LOBO", "ONÇA"],
    "Comida e Bebida": ["BOLO", "REFRIGERANTE", "BRIGADEIRO", "LASANHA", "MAÇÃ", "BANANA", "ABACATE", "LARANJA", "MORANGO", "ESPAGUETE", "ESPUMANTE"],
    "Roupas": ["CAMISETA", "CALÇA", "SAIA", "VESTIDO", "SAPATO", "BERMUDA", "BLUSA"],
    "Ciência e Natureza": ["FLORESTA", "UNIVERSO", "FISICA", "MEDICINA", "BIOLOGIA", "GENES", "JARDIM"],
    "Ações e Cotidiano": ["ESTUDAR", "TRABALHAR", "PROFESSORES", "FACA", "GARFO", "COLHER", "FOGUETE", "HIDRATANTE", "PAINEL", "CONGRESSO", "JOGAR", "ABRAÇAR"]
}

def escolher_categoria():
    print("\n--- ESCOLHA UMA CATEGORIA ---")
    lista_nomes = list(categorias.keys())

    # Exibe as opções numeradas
    for i, nome in enumerate(lista_nomes, 1):
        print(f"[{i}] {nome}")

    while True:
        escolha = input("\nDigite o número da categoria desejada: ").strip()
        if escolha.isdigit():
            indice = int(escolha) - 1
            if 0 <= indice < len(lista_nomes):
                categoria_selecionada = lista_nomes[indice]
                return categoria_selecionada, categorias[categoria_selecionada]
        print("Opção inválida! Escolha um número da lista.")

def jogar_forca():
    print(BANNER)
    print(DEMO_TAG)

    nome_cat, lista_palavras = escolher_categoria()
    palavra_secreta = random.choice(lista_palavras).upper()

    letras_descobertas = ["_"] * len(palavra_secreta)
    letras_tentadas = set()
    tentativas_restantes = 6

    print(f"\nCategoria selecionada: {nome_cat}")

    while tentativas_restantes > 0 and "_" in letras_descobertas:
        print("\nPalavra: " + " ".join(letras_descobertas))
        print(f"Tentativas restantes: {tentativas_restantes}")
        print(f"Letras tentadas: {', '.join(sorted(letras_tentadas)) if letras_tentadas else 'Nenhuma'}")

        palpite = input("Digite uma letra: ").strip().upper()

        if len(palpite) != 1 or not palpite.isalpha():
            print("Entrada inválida! Digite apenas uma letra.")
            continue

        if palpite in letras_tentadas:
            print("Você já tentou essa letra. Escolha outra.")
            continue

        letras_tentadas.add(palpite)

        if palpite in palavra_secreta:
            print(f"Boa! A letra '{palpite}' está na palavra.")
            for indice, letra in enumerate(palavra_secreta):
                if letra == palpite:
                    letras_descobertas[indice] = palpite
        else:
            print(f"Que pena! A letra '{palpite}' não está na palavra.")
            tentativas_restantes -= 1

    if "_" not in letras_descobertas:
        print("\n" + " ".join(letras_descobertas))
        print("\n🎉 PARABÉNS! Você venceu a demonstração!")
    else:
        print(f"\n💀 Fim de jogo na demonstração! A palavra era: {palavra_secreta}")

if __name__ == "__main__":
    jogar_forca()
```
