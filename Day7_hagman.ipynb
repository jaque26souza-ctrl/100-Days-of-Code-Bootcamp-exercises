#Jogo da Forca - Hangman

```python
# @title 🎮 CLIQUE AQUI PARA JOGAR

BANNER = """
====================================================================
                        J O G O   D A   F O R C A
====================================================================
"""
import random

#Lista de palavras
palavras = ["DESENVOLVIMENTO", "PYTHON", "ALGORITMO", "COMPUTADOR","FLORESTA", "UNIVERSO",
            "FISICA", "MEDICINA", "BIOLOGIA", "JARDIM", "GATO", "CACHORRO", "LAGARTO",
            "ESTUDAR", "TRABALHAR", "PROFESSORES", "BORBOLETA", "FACA", "GARFO",
            "COLHER", "BOLO", "REFRIGERANTE", "BRIGADEIRO"]

def jogar_forca():
    print(BANNER)
    palavra_secreta = random.choice(palavras).upper()
    letras_descobertas = ["_"] * len(palavra_secreta)
    letras_tentadas = set()
    tentativas_restantes = 6

    while tentativas_restantes > 0 and "_" in letras_descobertas:
        print("\nPalavra: " + " ".join(letras_descobertas))
        print(f"Tentativas restantes: {tentativas_restantes}")
        print(f"Letras tentadas: {', '.join(sorted(letras_tentadas)) if letras_tentadas else 'Nenhuma'}")

        palpite = input("Digite uma letra: ").strip().upper()

        if len(palpite) != 1 or not palpite.isalpha():
            print("Entrada inválida! Digite apenas uma letra.")
            continue

        if palpite in letras_tentadas:
            print("Você já tentou essa letra. Escolha outra.")
            continue

        letras_tentadas.add(palpite)

        if palpite in palavra_secreta:
            print(f"Boa! A letra '{palpite}' está na palavra.")
            for indice, letra in enumerate(palavra_secreta):
                if letra == palpite:
                    letras_descobertas[indice] = palpite
        else:
            print(f"Que pena! A letra '{palpite}' não está na palavra.")
            tentativas_restantes -= 1

    if "_" not in letras_descobertas:
        print("\n" + " ".join(letras_descobertas))
        print("\n🎉 PARABÉNS! Você venceu a demonstração!")
    else:
        print(f"\n💀 Fim de jogo na demonstração! A palavra era: {palavra_secreta}")

if __name__ == "__main__":
    jogar_forca()
```
