# Day 2: creating a bill calculator

### The calculator can calculate the percentage of the bill and how much each person should pay.


```python
print("Welcome to the bill calculator!")
bill = float(input("What was the total bill? $"))
tip = int(input("What percentage tip would you like to give? 10 12 15 "))
people = int(input("How many people to split the bill? "))
porcentage = bill * (tip/100)
total = bill + porcentage
total_person = total / people
print(f"Each person should pay: ${total_person:.2f}")
```
